```python
import pandas as pd
from collections import Counter
from sklearn.model_selection import train_test_split

# ===============================
# MEMBACA DATASET
# ===============================

df = pd.read_excel("../dataset/label_manual.xlsx")

X_train, X_test, y_train, y_test = train_test_split(
    df["komentar"],
    df["sentimen"],
    test_size=0.2,
    random_state=42,
    stratify=df["sentimen"]
)

# ===============================
# MEMISAHKAN DATA BERDASARKAN KELAS
# ===============================

data_pos = X_train[y_train=="positif"]
data_net = X_train[y_train=="netral"]
data_neg = X_train[y_train=="negatif"]

counter_pos = Counter()
counter_net = Counter()
counter_neg = Counter()

for kalimat in data_pos:
    counter_pos.update(kalimat.split())

for kalimat in data_net:
    counter_net.update(kalimat.split())

for kalimat in data_neg:
    counter_neg.update(kalimat.split())

vocabulary = (
    set(counter_pos.keys()) |
    set(counter_net.keys()) |
    set(counter_neg.keys())
)

print("="*60)
print("INFORMASI DATA TRAINING")
print("="*60)

print("Jumlah Data Training :", len(X_train))
print()

print(y_train.value_counts())
print()

print("Jumlah kata Positif :", sum(counter_pos.values()))
print("Jumlah kata Netral  :", sum(counter_net.values()))
print("Jumlah kata Negatif :", sum(counter_neg.values()))
print("Jumlah Vocabulary   :", len(vocabulary))

print()
print("="*60)
print("FREKUENSI KATA")
print("="*60)

for kata in ["herdman","out"]:
    print(
        f"{kata:10}"
        f" Positif={counter_pos[kata]:3}"
        f" Netral={counter_net[kata]:3}"
        f" Negatif={counter_neg[kata]:3}"
    )
```

```python
import joblib

# Load model dan TF-IDF yang sudah dilatih
model = joblib.load("../model/model_naive_bayes.pkl")
vectorizer = joblib.load("../model/tfidf.pkl")

komentar = ["herdman out"]

X = vectorizer.transform(komentar)

print("="*60)
print("HASIL PREDIKSI MODEL")
print("="*60)

prediksi = model.predict(X)[0]
probabilitas = model.predict_proba(X)[0]

print("Komentar :", komentar[0])
print("Prediksi :", prediksi)
print()

print("Probabilitas:")
for kelas, nilai in zip(model.classes_, probabilitas):
    print(f"{kelas:10}: {nilai:.6f}")
```
